# Load new data with explicit wavelength calibration

CW must be set explicitly. Typical workflow:
1. Load FITS with `Vis133M.from_files("file.fits")` (no CW applied yet).
2. Pick a frame with signal and plot spectrum in pixel space.
3. Estimate CW from a known line (e.g. H_gamma) at a chosen peak pixel.
4. Apply CW and plot calibrated spectrum.

```python
from bh_molecule.instruments import Vis133M

s = Vis133M.from_files("193791.fits")
```

```python
frame = s.pick_frame()
fig = s.plot_spectrum_pixels(frame, show_peaks=True)
fig.show()
```

```python
peak = 1917
cw = s.estimate_cw_from_line(peak_pixel=peak, line="H_gamma")
s.apply_cw(cw, line_name="H_gamma", peak_pixel=peak, frame_used=frame)
s.calibration_report()
```

```python
fig = s.plot_spectrum_plotly(10, 30)
fig.show()
```
