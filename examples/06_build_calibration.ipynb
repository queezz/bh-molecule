# Build Vis133M wavelength calibration

This notebook demonstrates how to build the **instrument** wavelength
calibration for the Vis133M spectrometer once, store it in `bh_wavecal.json`,
and then reuse it at runtime without reading the original instrument CSV.

There are two distinct types of CSV files in `bh_molecule._resources`:

1. `133mVis_wavcal.csv` – **instrument wavelength calibration** (dispersion
   vs pixel and channel). This is the *only* CSV that should be used to build
   `bh_wavecal.json`.
2. `11BH_v00.csv`, `11BH_v11.csv`, `11BH_v22.csv` – **BH molecular line
   constants** used for modelling the BH Q-branch. These are **not** wavelength
   calibration files and must never be used as wavecal input.

In this notebook we will:

1. Choose the instrument wavecal CSV (`133mVis_wavcal.csv`) and a reference
   Vis133M FITS cube.
2. Use `bh_molecule.calibration_builder.build_bh_wavecal` to compute the
   pixel→wavelength polynomial and write `bh_wavecal.json`.
3. Inspect the stored calibration using `load_bh_wavecal_json`.

> Note: you only need to run this notebook when you change the reference
> *instrument* calibration; normal analysis notebooks only need the generated
> `bh_wavecal.json`. 


```python
from bh_molecule.calibration_builder import build_bh_wavecal
from bh_molecule.instruments import load_bh_wavecal_json

# Reference calibration inputs
# wavcal CSV is resolved either from the filesystem or from the
# bh_molecule._resources package directory.
WAVCAL_CSV = "133mVis_wavcal.csv"  # Vis133M instrument wavelength calibration CSV
FITS = "./133mVis_169626.fits"      # update to your reference Vis133M cube
```

```python
# Build (or rebuild) the wavelength calibration JSON
params = build_bh_wavecal(
    wavcal_csv=WAVCAL_CSV,
    fits=FITS,
    channel=0,
    degree=2,
)
params
```

```python
# Inspect the stored calibration from bh_wavecal.json
cfg = load_bh_wavecal_json()
cfg
```
